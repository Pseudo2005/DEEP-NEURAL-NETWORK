```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay,f1_score
import tensorflow as tf
from sklearn.preprocessing import LabelEncoder,StandardScaler
import torch
import torch.nn as nn
from torchsummary import summary
```

```python
RANDOM_STATE = 24  # Remember to remove at the time of promotion to production(RANDOM_STATE = None) 
np.random.seed(RANDOM_STATE)
rng = np.random.default_rng(seed = RANDOM_STATE)

NOISE = 0.2
N_EPOCHS = 1000 # Number of Epochs
ALPHA = .001  # Learning Rate 
N_SAMPLES = 1000  # number of sample to create
TEST_SIZE = 0.2

params = {
    'legend.fontsize':'medium',
    'figure.figsize':(15,6),
    'axes.labelsize':'medium',
    'axes.titlesize':'medium',
    'xtick.labelsize':'medium',
    'ytick.labelsize':'medium'
}
plt.rcParams.update(params)
CMAP = plt.cm.coolwarm
plt.style.use('seaborn-v0_8-darkgrid') 
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using {torch.cuda.get_device_name(0) if device.type == 'cuda' else 'cpu'}")
```

```python
data_df=pd.read_csv("wine.csv",header=None)
data_df.head()
```

```python
labels=data_df[data_df.columns[-1]]
features_df=data_df.drop(data_df.columns[-1],axis=1)
features_df.shape,labels.shape
```

```python
X_train,X_test,y_train,y_test=train_test_split(features_df,
                                               labels,
                                               test_size=TEST_SIZE,
                                               random_state=RANDOM_STATE,
                                               stratify=labels)
```

```python
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
le=LabelEncoder()
y_train=le.fit_transform(y_train)
y_test=le.transform(y_test)
```

# prepare Model

```python
model=nn.Sequential(
    nn.Linear(13,8),
    nn.ReLU(),
    nn.Linear(8,3),)
display(model)
model=model.to(device)
```

```python
print(f'Num Parameter {len(list(model.parameters()))}')
print(f'Layer 1 weight{list(model.parameters())[0].shape}')
print(f'Layer 1 bais{list(model.parameters())[1].shape}')
print(f'Layer 2 weight{list(model.parameters())[2].shape}')
print(f'Layer 1 bais{list(model.parameters())[3].shape}')
```

```python
loss_fn=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters()      #parameter to optimize
                           ,lr=ALPHA)              #Learning rate
```

```python
train_x = torch.tensor(X_train, dtype=torch.float32, device=device)
train_y = torch.tensor(y_train, dtype=torch.int64, device=device)

test_x = torch.tensor(X_test, dtype=torch.float32, device=device)
test_y = torch.tensor(y_test, dtype=torch.int64, device=device)
```

```python
train_loss,test_loss=[],[]
for epoch in range(N_EPOCHS):
    model.train()
    outputs=model(train_x)  #forward propogation
    loss =loss_fn(outputs, train_y)
    optimizer.zero_grad()   #reset
    loss.backward()         #back prop
    optimizer.step()        #W=W+alpha* dW
    train_loss.append(loss.item())

    with torch.inference_mode():
        model.eval()
        outputs=model(test_x)   #forward propagation
        tloss=loss_fn(outputs,test_y)  #calculate loss
        test_loss.append(tloss.item())
    
    if epoch % 100==0:
        print(f'at epoch-{epoch:4d}: Loss -{loss.item():.4f}')
```

```python
fig ,ax=plt.subplots(figsize=(12,4))
ax.plot(train_loss,label='Train')
ax.plot(test_loss,label='Test')
ax.legend()
```

```python
model.eval()
with torch.inference_mode():
    train_pred=model(train_x)
    test_pred=model(test_x)
```

```python
train_pred.shape
```

```python
train_pred.argmax(dim=1)
```

```python
y_train_pred=train_pred.argmax(dim=1).detach()  #remove computation graph
y_train_pred=y_train_pred.cpu().numpy()
y_train_pred
```

```python
accuracy_score(y_train,y_train_pred)
```

```python
f_score=f1_score(y_true=y_train,y_pred=y_train_pred,average='weighted')
accuracy_score(y_train,y_train_pred)
```

```python
cm = confusion_matrix(y_train, y_train_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[1,2,3])
fig, ax = plt.subplots(figsize=(4,4))
disp.plot(ax=ax, cmap='Blues', colorbar=False)
ax.set_title(f'F1 Score: {f_score:0.3f}')
```

## for test

```python
y_test_pred=test_pred.argmax(dim=1).detach()  #remove computation graph
y_test_pred=y_test_pred.cpu().numpy()
y_test_pred
```

```python
f_score=f1_score(y_true=y_test,y_pred=y_test_pred,average='weighted')
accuracy_score(y_test,y_test_pred)
```

```python
cm = confusion_matrix(y_test, y_test_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[1,2,3])
fig, ax = plt.subplots(figsize=(4,4))
disp.plot(ax=ax, cmap='Blues', colorbar=False)
ax.set_title(f'F1 Score: {f_score:0.3f}')
```
